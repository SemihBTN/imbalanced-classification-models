```python
def correlation_for_dropping(df, threshold):
    columns_to_drop = set()
    corr = df.corr(numeric_only=True)
    for i in range(len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i, j]) > threshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X = df.drop("Machine failure",axis=1)
y= df["Machine failure"]
```

```python
X_train , X_test , y_train , y_test = train_test_split(X,y,test_size=0.25,random_state=15)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
```

```python
param_grid_lr = {
    'penalty': ['l1', 'l2'],
    'C': [0.01, 0.1, 1.0, 10],
    'solver': ['liblinear', 'saga'],
    'class_weight': ['balanced', {0: 1, 1: 10}, {0: 1, 1: 50}]
}
```

```python
logreg = LogisticRegression()
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
grid = GridSearchCV(estimator=logreg, param_grid=param_grid_lr, scoring='f1', cv=cv, n_jobs=-1)
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(X_train_scaled,y_train)
```

```python
grid.best_params_
```

```python
y_pred=grid.predict(X_test_scaled)
```

```python
from sklearn.metrics import roc_auc_score, roc_curve, classification_report, confusion_matrix
import matplotlib.pyplot as plt

# 1. Test verisinin pozitif sınıf olma olasılıklarını alıyoruz
model_prob = grid.predict_proba(X_test_scaled)[:, 1]

# 2. ROC-AUC Skoru
model_auc = roc_auc_score(y_test, model_prob)
print(f"Model AUC Skoru: {model_auc:.4f}")

# 3. ROC Eğrisi için FPR, TPR ve eşikleri hesaplıyoruz
model_fpr, model_tpr, thresholds = roc_curve(y_test, model_prob)

# ROC Eğrisini Çizdirme
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(model_fpr, model_tpr, marker='.', label='Logistic Regression')

# Grafiğe eşik değerlerini küçük etiketler olarak eklemek istersen:
# for fpr, tpr, thresh in zip(model_fpr, model_tpr, thresholds):
#     ax.annotate(f'{np.round(thresh, 2)}', (fpr, tpr), textcoords="offset points", xytext=(0,10), ha='center', fontsize=8)

ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.set_title('ROC Curve - Logistic Regression')
ax.legend()
ax.grid(True)
plt.show()

# 4. Kendi belirlediğimiz özel eşik değeri (Örn: 0.23)
custom_threshold = 0.45

y_pred_custom = (model_prob >= custom_threshold).astype(int)

# 5. Sonuçları Değerlendirme
print(f"\nEvaluation using custom threshold = {custom_threshold}")
print(confusion_matrix(y_test, y_pred_custom))
print("\n")
print(classification_report(y_test, y_pred_custom))
```
